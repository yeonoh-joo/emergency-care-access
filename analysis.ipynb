```python
# 1

# 라이브러리 불러오기
import pandas as pd
from os import getenv
from dotenv import load_dotenv
import numpy as np
import requests, xmltodict, json
```

```python
# 2

# 응급의료기관 정보 API 주소
url = 'https://apis.data.go.kr/B552657/ErmctInfoInqireService/getEgytListInfoInqire'

# 환경변수(.env)에서 API KEY 가져오기
load_dotenv()
key = getenv('d_key')

# 분석 대상 지역 (수도권 3개 시도)
regions = ['서울특별시', '경기도', '인천광역시']
rows = []

# 지역별 API 호출 반복
for region in regions:

    # API 요청 파라미터 세팅
    param = {
    'serviceKey': key,
    'Q0': region,  # 시도 기준으로 필터링
    'numOfRows': 1000  # 충분한 수의 병원 데이터 확보
    }

    # XML → JSON → Python dict 변환
    res = requests.get(url, params=param).text
    data = json.loads(json.dumps(xmltodict.parse(res)))['response']['body']['items']['item']

    # 한 지역의 모든 병원 정보 저장
    for i, v in enumerate(data):
        name = v['dutyName']
        tel = v['dutyTel1']
        addr = v['dutyAddr']
        lati = v['wgs84Lat']
        longi = v['wgs84Lon']

        # 데이터가 제대로 나오는지 확인
        print(f'{i+1}\t{name}\t{tel}\t{addr}\t{lati}\t{longi}')

        # 데이터프레임 생성용 리스트에 추가
        rows.append({
            '시도': region,
            '병원명': name,
            '전화번호': tel,
            '주소': addr,
            '위도': lati,
            '경도': longi
        })

# 수집된 병원 수 확인
print('수집된 병원 수: ', len(rows))
```

```python
# 3

# 리스트 → DataFrame 변환 (병원 데이터 완성)
df_hosp = pd.DataFrame(rows)
df_hosp.head(20)
```

```python
# 4

# 지역별 병원 수 확인 (데이터 정상 수집 여부 체크)
df_hosp['시도'].value_counts()
```

```python
# 5

# 주소에서 시군구 정보 추출 (공백 기준으로 두 번째 요소)
df_hosp['시군구'] = df_hosp['주소'].str.split().str[1]

# 시도 + 시군구 기준 응급의료기관 수 집계
hosp_by_region = df_hosp.groupby(['시도', '시군구']).size().reset_index(name='응급의료기관수')
hosp_by_region.head(15)
```

```python
# 6

# 지역별 인구 통계 데이터 불러오기
df_pop = pd.read_csv('행정안전부_지역별(행정동) 성별 연령별 주민등록 인구수.csv', encoding='cp949')

# 컬럼명 공백 제거 (오류 방지)
df_pop.columns = df_pop.columns.str.strip()

df_pop['인구수'] = df_pop['계']

# 65세 이상 인구 계산
senior_m = [col for col in df_pop.columns if '세남자' in col and int(col.split('세')[0]) >= 65]
senior_f = [col for col in df_pop.columns if '세여자' in col and int(col.split('세')[0]) >= 65]
df_pop['65세 이상 인구수'] = df_pop[senior_m + senior_f].sum(axis=1)

# 필요한 지역만 추출
df_pop = df_pop[df_pop['시도명'].isin(regions)].copy()

# 시군구 이름 맞추기 (예: '수원시 장안구' -> '수원시')
df_pop['시군구'] = df_pop['시군구명'].apply(lambda x: x.split()[0])

# '시도'와 '시군구' 기준으로 인구 합치기 (동 단위 -> 구 단위 통합)
df_pop_final = df_pop.groupby(['시도명', '시군구'])[['인구수', '65세 이상 인구수']].sum().reset_index()

# 병합을 위해 컬럼명 변경 (시도명 -> 시도)
df_pop_final.rename(columns={'시도명': '시도'}, inplace=True)

df_pop_final.head(15)
```

```python
# 7

# 시도와 시군구 기준으로 병합
df_merge = pd.merge(df_pop_final, hosp_by_region, on=['시도', '시군구'], how='left')

# 병원이 없는 곳은 NaN 대신 0으로 채움
df_merge['응급의료기관수'] = df_merge['응급의료기관수'].fillna(0).astype(int)

df_merge.head(15)
```

```python
# 8

# 인구 10만명당 응급의료기관 수 계산
df_merge['인구 10만명당 기관수'] = df_merge['응급의료기관수'] / df_merge['인구수'] * 100000

# 인구 10만명당 응급의료기관 수가 0.6개 미만인 곳을 취약지로 선정
cutline = 0.6

df_merge['취약지역여부'] = df_merge['인구 10만명당 기관수'].apply(lambda x: 'O' if x < cutline else 'X')

# 취약지역 확인
risk = df_merge[df_merge['취약지역여부'] == 'O']
print(f"취약지역 수: {len(risk)}곳")

df_merge
```
